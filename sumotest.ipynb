```python
# List counts and sample IDs for nodes and edges
try:
    net  # noqa: F821
except NameError:
    import sumolib
    net = sumolib.net.readNet('trikala_maas_project/network/trikala_center_sidewalks.net.xml')

nodes = list(net.getNodes())
edges = list(net.getEdges())
print(f"Nodes: {len(nodes)}, Edges: {len(edges)}")
print("Sample node IDs:", [n.getID() for n in nodes[:10]])
print("Sample edge IDs:", [e.getID() for e in edges[:10]])
```

```python
# Helper search functions to find IDs

def find_nodes(substr=None):
    results = []
    for n in net.getNodes():
        nid = n.getID()
        if substr is None or substr in nid:
            results.append((nid, n.getCoord()))
    return results


def find_edges(substr=None, from_node=None, to_node=None):
    results = []
    for e in net.getEdges():
        eid = e.getID()
        if substr and substr not in eid:
            continue
        if from_node and e.getFromNode().getID() != from_node:
            continue
        if to_node and e.getToNode().getID() != to_node:
            continue
        results.append((eid, e.getFromNode().getID(), e.getToNode().getID()))
    return results

# Examples (adjust the substrings/IDs to your network)
print('First node coord example:')
first_node_id = next(iter(net.getNodes())).getID()
print(first_node_id, net.getNode(first_node_id).getCoord())

print('\nFind nodes containing "center":')
print(find_nodes('center')[:5])

print('\nFind edges from the first node:')
print(find_edges(from_node=first_node_id)[:5])
```
